```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import circlify

from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 10 suurinta maata
suurimmat = (df.nlargest(10, "2026")["name"])
df_valitut = df[df["name"].isin(suurimmat)].sort_values("2026", ascending=False)

years = range(1800, 2101, 2)

fig, ax = plt.subplots(figsize=(5,5))

def update(year):
    year = str(year)
    ax.clear()

    circles = circlify.circlify(
        df_valitut[year].tolist(),
        show_enclosure=False
    )
    circles = sorted(circles, key=lambda c: c.r, reverse=True)

    colors = cm.tab10(range(len(df_valitut)))

    for color, (circle, (_, row)) in zip(colors, zip(circles, df_valitut.iterrows())):
        ax.add_patch(

            plt.Circle(
                (circle.x, circle.y),
                circle.r,
                color=color,
                alpha=0.7
            )
        )

        fontsize = max(6, int(circle.r * 20))
        #print(row["name"], circle.r)
        ax.text(
            circle.x,
            circle.y,
            row["name"],
            ha="center",
            va="center",
            fontsize=fontsize
        )

    ax.set_title(f"Väkiluku vuonna {year}")
    ax.set_aspect("equal")
    ax.set_xlim(-1.1, 1.1)
    ax.set_ylim(-1.1, 1.1)
    ax.axis("off")

animaatio = FuncAnimation(
    fig,
    update,
    frames=years,
    interval=100,
    repeat=False
)

plt.close(fig)

HTML(animaatio.to_jshtml())

```
